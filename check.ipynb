```python
p10 = web.DataReader('10_Portfolios_Prior_12_2', 'famafrench',start=START)
p10[0]
```

```python
p10_mean = p10[0].mean()
p10_mean
```

```python
import matplotlib.pyplot as plt

mean_returns = p10[0].mean()

colors = ['#B4B2A9'] * 10
colors[0] = '#D85A30'
colors[9] = '#1D9E75'

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar(range(10), mean_returns.values, color=colors, edgecolor='black', linewidth=0.5)
ax.set_xticks(range(10))
ax.set_xticklabels(mean_returns.index, rotation=45, ha='right')
ax.set_xlabel('Momentum decile')
ax.set_ylabel('Mean monthly return (%)')
ax.set_title('Mean monthly return by momentum decile, 1963-2026')
ax.axhline(0, color='black', linewidth=0.8)
fig.tight_layout()
fig.savefig('figures/momentum_deciles.png', dpi=150)
plt.show()
```
